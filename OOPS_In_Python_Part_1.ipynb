```python
class TradingStrategy:
  def __init__(self,symbol,amount):
    self.symbol=symbol
    self.amount=amount
    self.position=0

  def Buy(self,price,quantity):
    if self.amount>=price*quantity:
      self.position+=quantity
      self.amount-=price*quantity
      print(f"BUY Successfully {quantity} shares of {self.symbol} at {price}")

    else:
      print("Insufficient funds")

  def Sell(self,price,quantity):
    if self.position>=quantity:
      self.position-=quantity
      self.amount+=price*quantity
      print(f"SELL Successfully {quantity} shares of {self.symbol} at {price}")
    else:
      print("no share to sell")
  def Status(self):
    print(f"Current Position {self.position} of shares")
    print(f"Remaing amount {self.amount}")
```

```python
momentum=TradingStrategy("BTC",5000)
```

```python
momentum.amount
```

```python
momentum.symbol
```

```python
momentum.Buy(1200,3)
```

```python
momentum.Status()
```

```python
momentum.Sell(1300,2)
```
