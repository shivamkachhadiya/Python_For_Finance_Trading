```python
class AutoTradingBot:
  def __init__(self,thresold):
    self.__thresold=thresold
    self.__position=None
    self.__price_data=[]

  def fetch_market_data(self):
    self.__price_data=[69160,67140,69240,69120,69740,69540]
    return self.__price_data
    print("Market data has been fetched..")


  def letest_price(self):
    if self.__price_data:
      return self.__price_data[-1]
    else:
      None


  def evaluate_price(self):
    latest_price=self.letest_price()
    if latest_price<self.__thresold:
      return "SELL"
    elif latest_price>self.__thresold:
      return "BUY"
    else:
      return "HOLD"

  def execute_trade(self):
    if self.evaluate_price()=="SELL" and self.__position!="SHORT":
      self.__position="SHORT"
      return "SELL TRADE EXECUTED & CURRENT POSITION IS SHORT"

    elif self.evaluate_price()=="BUY" and self.__position!="LONG":
      self.__position="SHORT"
      return "BUY TRADE EXECUTED & CURRENT POSITION IS LONG"

    elif self.evaluate_price()=="HOLD":
      self.__position="HOLD"
      return "HOLD TRADE EXECUTED & CURRENT POSITION IS HOLD"



  def run():
    pass

```

GETTER AND SETTER


```python
class AutoTradingBot:
  def __init__(self,thresold):
    self.__thresold=thresold
    self.__position=None
    self.__price_data=[]

  def getattr(self):
    return self._AutoTradingBot__thresold

  def setattr(self,new_thresold):
    self._AutoTradingBot__thresold=new_thresold
    return self._AutoTradingBot__thresold
```
