```python

import yfinance as yf
import numpy as np

class QuantMPT:
    def __init__(self, tickers):
        self.tickers = tickers
        self.data = None

    def fetch_data(self, period="1y"):
        print("Fetching data...")
        self.data = yf.download(
            self.tickers,
            period=period,
            auto_adjust=True
        )["Close"]
        return self.data

    def calculate_mpt_metrics(self):
        if self.data is None:
            self.fetch_data()

        returns = self.data.pct_change().dropna()

        mean_returns = returns.mean() * 252
        volatility = returns.std() * np.sqrt(252)
        corr_matrix = returns.corr()

        return {
            "Expected Returns (%)": (mean_returns * 100).round(2),
            "Volatility (%)": (volatility * 100).round(2),
            "Correlation Matrix": corr_matrix.round(2)
        }


```

```python
# Testing MPT
stocks = ["AAPL", "MSFT", "GOOGL"]

mpt = QuantMPT(stocks)
metrics = mpt.calculate_mpt_metrics()

print("\n--- MPT Portfolio Statistics ---")

print("\n1. Expected Annual Returns:")
print(metrics["Expected Returns (%)"])

print("\n2. Annualized Volatility (Risk):")
print(metrics["Volatility (%)"])

print("\n3. Correlation Matrix:")
print(metrics["Correlation Matrix"])
```
