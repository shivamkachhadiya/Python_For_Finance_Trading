```python
import json
L=[
    {
        "tradeId":"T3214","Symbol":"APLE","price":100,"Volume":123
    },
    {
        "tradeId":"T3254","Symbol":"MSFT","price":140,"Volume":723
    }
]
with open("trade_log.json","w")as f:
  json.dump(L,f)
```

```python
with open("trade_log.json","r")as f:
  print(f.read())
```

```python
with open("trade_log.json","r")as f:
  print(json.load(f))
```

```python
t=("t123","AA",100)
with open("touple.json","w")as f:
  json.dump(t,f)
```

```python
with open("touple.json","r")as f:
  print(json.load(f))
```
