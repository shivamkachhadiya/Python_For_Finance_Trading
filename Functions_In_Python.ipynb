```python
def cal_sma(prices,window):
  '''
  it will calculate moving avrage
  parameter list of price and window size
  return moving average
  '''
  if len(prices)<window:
    print("value error...")

  sum_of_prices=sum(prices[-window:])
  sma=sum_of_prices/window
  return sma
```

```python
prices=[10,20,30,40,50,60,20,304,50,403,40]
print("simple moving avrage is ",cal_sma(prices,5))
```
