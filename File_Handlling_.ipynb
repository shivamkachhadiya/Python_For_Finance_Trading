```python
f=open("traders.csv","r")
while True:
  data=f.readline()
  if data=='':
    break
  else:
    print(data,end='')
f.close()
```

```python
import csv
with open("traders.csv","w",newline='')as f:
  writer=csv.writer(f)
  writer.writerow(["symbol","price","qty"])
  writer.writerow(["AAPL","1222","6"])
  writer.writerow(["MSFT","200","90"])
```
