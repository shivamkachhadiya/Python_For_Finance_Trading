```python
class MyTradingBot:
  def __init__(self,name,strategy):
    self.name=name
    self.strategy=strategy

  def Execute(self,price):
    action=self.strategy.Execute(price)
    return f"{self.name} is execting a {action} action"

class AwosomeStrategy:
  def __init__(self,thresold):
    self.thresold=thresold

  def Execute(self,latest_price):
    if latest_price>self.thresold:
      return "BUY"
    else:
      return "SELL"
```

```python
strategy=AwosomeStrategy(100)
mybot=MyTradingBot("AwosomeStrategy",strategy)
mybot.Execute(190)
```
