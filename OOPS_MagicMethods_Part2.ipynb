```python
class Trades:
  def __init__(self,symbol,quantity,price):
    self.symbol=symbol
    self.quantity=quantity
    self.price=price
    print(" I am DONE")

  def __repr__(self):
    return f"Symbol={self.symbol},quantity={self.quantity},price={self.price}"

  def __str__(self):
    return f"{self.quantity}shares of{self.symbol} at {self.price}"

  def __add__(self,other):
    if self.symbol==other.symbol:
      new_quantity=self.quantity+other.quantity
      avg_price=(self.quantity*self.price+other.quantity*other.price)/new_quantity
      return f"The latest quantity is {new_quantity}with an avrage price of {avg_price}"
```

```python
trade1=Trades("APLE",3,33)
```

```python
print(repr(trade1))
```

```python
print(str(trade1))
```
